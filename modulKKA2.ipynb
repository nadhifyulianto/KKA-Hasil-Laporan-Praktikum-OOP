```python
import numpy as np
harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

Apa yang terjadi jika mencoba [5000, 7000, 3000] * 0.9 pada List Python biasa?
Akan menghasilkan TypeError: can't multiply sequence by non-int of type 'float'.

```python
import pandas as pd
data_kantin = {
'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
'harga': [12000, 4000, 10000, 4000, 8000],
'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

Kolom dengan data kosong (None/NaN):

Kolom terjual pada indeks ke-2 (bernilai NaN).

Kolom menu pada indeks ke-4 (bernilai None).

```python
df = pd.read_csv('data_kantin.csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Analisis df.info():Kolom yang memiliki jumlah non-null lebih sedikit dibanding total baris adalah kolom menu (4 non-null) dan kolom terjual (4 non-null) dari total 5 baris data.  Artinya:
Terdapat missing values (nilai kosong/hilang) pada kedua kolom tersebut. Ada 1 baris yang tidak memiliki nilai menu, dan 1 baris yang tidak memiliki nilai

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Kode tersebut digunakan untuk membuat dan mengolah dataset kantin menggunakan Pandas. Dataset berisi nama menu, harga, dan jumlah yang terjual. pd.DataFrame() digunakan untuk mengubah data menjadi tabel, sedangkan print() digunakan untuk menampilkan data. isnull().sum() digunakan untuk mengecek jumlah data yang kosong pada setiap kolom.

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Jumlah baris setelah drop_duplicates():
Sebelumnya terdapat 4 baris data tersisa (setelah dropna pada Latihan 4). Karena 'Es Teh' muncul 2 kali dengan data yang sama persis (harga: 4000, terjual: 40), 1 baris duplikat dihapus, sehingga jumlah baris dataset menjadi 3 baris.

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

Menu yang menghasilkan total_pendapatan tertinggi:
Nasi Goreng dengan total pendapatan sebesar Rp276.000.
